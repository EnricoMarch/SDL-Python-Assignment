Modify to allow user to input the value and word manually

```python
%%writefile fizz_mod.py

import sys

def print_selection(dict):
    for i in range(1,101):
        line=""
        for key, value in dict.items():
            if i % key == 0:
                line += value
        if line:
            print(line)
        else:
            print(i)


def main(num, word):
    num = int(num)
    dictionary = {num: word}
    print_selection(dictionary)

if __name__ == '__main__':
    num = sys.argv[1]
    word = sys.argv[2]
    main(num, word)
```

Modification to specify max number

```python
%%writefile fizz_mod.py

import sys

def print_selection(dict, max):
    for i in range(1,max+1):
        line=""
        for key, value in dict.items():
            if i % key == 0:
                line += value
        if line:
            print(line)
        else:
            print(i)


def main(num, word, max):
    num = int(num)
    max = int(max)
    dictionary = {num: word}
    print_selection(dictionary, max)

if __name__ == '__main__':
    num = sys.argv[1]
    word = sys.argv[2]
    max = sys.argv[3]
    main(num, word, max)
```
